```python
import numpy as np 
import pandas as pd 
from mlxtend.frequent_patterns import apriori, association_rules 
```

```python
data = pd.read_csv('/kaggle/input/combine-dataset-2009-2011/combine_dataset_2009_2011.csv') 
data.head() 
```

```python
data.columns 
```

```python
data.Country.unique() 
```

```python
data['Description'] = data['Description'].str.strip() 
  
# Dropping the rows without any invoice number 
data.dropna(axis = 0, subset =['Invoice'], inplace = True) 
data['Invoice'] = data['Invoice'].astype('str') 
  
# Dropping all transactions which were done on credit 
data = data[~data['Invoice'].str.contains('C')] 
```

```python
# Transactions done in the United Kingdom  
basket=data.groupby(['Invoice', 'Description'])['Quantity'] .sum().unstack().reset_index().fillna(0).set_index('Invoice') 
```

```python
def hot_encode(x): 
    if(x<= 0): 
        return 0
    if(x>= 1): 
        return 1
basket_encoded = basket.applymap(hot_encode) 
basket= basket_encoded 
```

```python
# pip install --upgrade mlxtend
```

```python
# Generate frequent itemsets
frq_items = apriori(basket, min_support=0.02, use_colnames=True)

# Generate association rules

rules = association_rules(frq_items, metric="lift", min_threshold=1, support_only=False,num_itemsets=len(frq_items))

# Sort rules by confidence and lift
rules = rules.sort_values(['confidence', 'lift'], ascending=[False, False])

# Print the top rules
print(rules.head()[["antecedents","consequents", "antecedent support","consequent support"]])
```
